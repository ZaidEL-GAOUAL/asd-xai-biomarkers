                # Step 1 - Data Loading and Metadata

                This notebook isolates the first stage of the ASD transcriptomics workflow.
                It downloads or reloads `GSE18123`, extracts the diagnosis labels and platform
                identifiers, and saves a clean sample metadata table for the next steps.
                


```python
import sys; sys.path.insert(0, '../src')
                import pandas as pd

                from asd_pipeline_utils import (
                    DESTDIR,
                    GSE_ID,
                    METADATA_PATH,
                    extract_sample_metadata,
                    load_geo_series,
                    save_metadata,
                )

                # Load the GEO series and recover the raw probe-by-sample matrix.
                gse, expr_probes = load_geo_series(GSE_ID, DESTDIR)

                # Extract the diagnosis and platform metadata that will drive every later stage.
                expr_probes, diagnosis_df, platform_s = extract_sample_metadata(gse, expr_probes)
                metadata_df = save_metadata(diagnosis_df, platform_s, METADATA_PATH)

                print(f"Saved metadata to: {METADATA_PATH.resolve()}")
                metadata_df.head()
                
```

```python
                # Inspect the basic class and platform composition before any harmonization step.
                print("Platform counts:")
                print(platform_s.value_counts())

                print("\nDiagnosis counts:")
                print(metadata_df["diagnosis"].value_counts())

                print("\nDiagnosis by platform:")
                print(pd.crosstab(metadata_df["diagnosis"], metadata_df["platform_id"]))
                
```
